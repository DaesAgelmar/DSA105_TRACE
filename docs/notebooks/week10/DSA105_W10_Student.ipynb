## Part A — Core understanding

### 1. First reusable function


```python
def calculate_revenue(quantity, unit_price):
    """Return revenue from quantity and unit price."""
    return quantity * unit_price


revenue = calculate_revenue(12, 50)
print("Revenue:", revenue)
```

Why is `calculate_revenue` clearer than `calc`? Why is the result returned rather than printed inside the function?


### 2. Parameter variation


```python
def apply_discount(price, discount_rate):
    """Return price after applying a proportional discount."""
    return price * (1 - discount_rate)


print(apply_discount(500, 0.10))
print(apply_discount(500, 0.25))
```

Change one argument at a time and explain which dependency the parameter represents.


### 3. One responsibility per function


```python
def calculate_profit(revenue, cost):
    """Return profit from revenue and cost."""
    return revenue - cost
```

The function calculates. The surrounding program can decide whether to print, store, compare, or reuse the result.
