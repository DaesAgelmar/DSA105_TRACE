## 2. Algorithm = A Precise Sequence

An **algorithm** is a finite and unambiguous sequence of steps for solving a problem.

A useful beginner checklist:

- **Input:** What information enters the solution?
- **Process:** What operations transform the information?
- **Output:** What result must be produced?
- **Order:** In which sequence must the steps happen?
- **Termination:** Does the procedure end?
