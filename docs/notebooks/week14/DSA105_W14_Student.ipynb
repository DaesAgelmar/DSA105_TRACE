### 1. Decompose before coding

Problem:

> A shop has transaction prices. Prices strictly above a threshold receive a configurable percentage discount. Calculate the average adjusted price.

### 2. Reusable transformations


```python
def apply_discount(price, threshold, discount_rate):
    """Return price after applying a discount above the threshold."""
    if price > threshold:
        return price * (1 - discount_rate)

    return price


def calculate_average(values):
    """Return the arithmetic mean of a non-empty list."""
    total = 0

    for value in values:
        total += value

    return total / len(values)
```

### 3. Compose the functions


```python
prices = [80, 120, 200, 100]
threshold = 100
discount_rate = 0.10

adjusted_prices = []

for price in prices:
    adjusted_price = apply_discount(
        price,
        threshold,
        discount_rate,
    )
    adjusted_prices.append(adjusted_price)

average_price = calculate_average(adjusted_prices)

print("Adjusted prices:", adjusted_prices)
print(f"Average adjusted price: {average_price:.2f}")
```

## Part B — Verification

### 4. AI Verify — two independent defects


```python
def average_adjusted_price(prices, threshold, discount_rate):
    """Return the average price after applying an eligible discount."""
    adjusted_prices = []

    for price in prices:
        if price > threshold:
            adjusted_prices.append(price - discount_rate)
        else:
            adjusted_prices.append(price)

    total_adjusted_price = 0

    for index in range(len(adjusted_prices) - 1):
        total_adjusted_price += adjusted_prices[index]

    return total_adjusted_price / len(adjusted_prices)


prices = [80, 120, 200, 100]
print(average_adjusted_price(prices, 100, 0.10))
```

Intended rule:

- only prices strictly above 100 are discounted;
- the discount is 10%, not 0.10 currency units;
- every adjusted price belongs in the average.

Identify exactly two independent defects:

1. rate meaning;
2. omitted final value.

Design a test that exposes each.
