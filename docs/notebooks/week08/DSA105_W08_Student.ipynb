### 2. Counter


```python
day_count = 0

for _ in range(4):
    day_count += 1

print("Days counted:", day_count)
```

`_` is appropriate here because the loop variable itself is not used.


### 3. Accumulator

Predict the final total before running.


```python
total_revenue = 0

for day in range(1, 5):
    daily_revenue = 100
    total_revenue += daily_revenue

print("Total revenue:", total_revenue)
```

In this example, `daily_revenue` is more informative than `i` because the loop iterates over revenue values, not positions.


### 4. Interactive repetition


```python
total_revenue = 0

for day in range(1, 5):
    daily_revenue = float(input(f"Revenue for day {day}: "))
    total_revenue += daily_revenue

average_revenue = total_revenue / 4

print(f"Four-day total: {total_revenue:.2f}")
print(f"Average daily revenue: {average_revenue:.2f}")
```
