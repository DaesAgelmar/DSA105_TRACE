### 2. Correct two-level accumulation


```python
company_total = 0

for branch in range(1, 3):
    branch_total = 0

    for day in range(1, 4):
        daily_sales = float(
            input(f"Branch {branch}, day {day} sales: ")
        )
        branch_total += daily_sales

    print(f"Branch {branch} total: {branch_total:.2f}")
    company_total += branch_total

print(f"Company total: {company_total:.2f}")
```

Explain:

- why `branch_total = 0` belongs inside the outer loop;
- why `company_total = 0` belongs before both loops.


## Part B — Practice and verification

### 3. Debug This — reset in the wrong place


```python
company_total = 0
branch_total = 0

for branch in range(1, 3):
    for day in range(1, 4):
        daily_sales = float(input("Sales: "))
        branch_total += daily_sales

    company_total += branch_total

print(f"Company total: {company_total:.2f}")
```

Use:

- Branch 1: `100, 120, 80`
- Branch 2: `90, 110, 130`

Predict `930` before running and explain exactly why the first branch becomes double-counted.


### 4. AI Verify — overwrite instead of accumulate


```python
company_total = 0

for branch in range(1, 3):
    branch_total = 0

    for day in range(1, 4):
        daily_sales = float(input("Sales: "))
        branch_total += daily_sales

    company_total = branch_total

print(f"Company total: {company_total:.2f}")
```

Correct total: `630`.

Predict → Run → Diagnose → Fix → Verify.

The important distinction:

- wrong reset location → state leaks and earlier values are counted again;
- assignment instead of accumulation → earlier state is overwritten.
