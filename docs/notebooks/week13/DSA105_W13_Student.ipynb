```python
def read_quantity():
    """Return an integer quantity entered by the user."""
    try:
        quantity = int(input("Quantity: "))
    except ValueError:
        return None

    return quantity
```

## Part B — Practice and verification

### 3. AI Verify — 'safe' code that hides too much


```python
def safe_divide(numerator, denominator):
    """Return a quotient while handling division by zero."""
    try:
        return numerator / denominator
    except:
        return 0


print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide(10, "two"))
```

Why is `0` a misleading result for the wrong-type call?


```python
def safe_divide(numerator, denominator):
    """Return a quotient while handling division by zero."""
    try:
        return numerator / denominator
    except ZeroDivisionError:
        return None
```

Now `safe_divide(10, "two")` remains visible as an unexpected error. That is useful information.
