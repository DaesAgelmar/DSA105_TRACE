## Part A — Core understanding

### 1. Inspect a list


```python
daily_sales = [120, 95, 130, 80]

print("First value:", daily_sales[0])
print("Number of values:", len(daily_sales))
```

### 2. Iterate directly over values


```python
total_sales = 0

for daily_sale in daily_sales:
    total_sales += daily_sale

print("Total sales:", total_sales)
```

If you do not need the position, `daily_sale` communicates more than an index such as `i`.


### 3. Build a filtered list


```python
threshold = 100
high_sales = []

for daily_sale in daily_sales:
    if daily_sale >= threshold:
        high_sales.append(daily_sale)

print(high_sales)
```

This pattern reads from one collection and creates another. It does not change the collection being traversed.


### 4. Optional compact form

Use only after you can explain the explicit loop.


```python
high_sales = [
    daily_sale
    for daily_sale in daily_sales
    if daily_sale >= threshold
]

print(high_sales)
```
