## 1. Inspect a list


```python
sales = [120, 95, 130, 80]
print(sales[0])
print(len(sales))
```

## 2. Aggregate list values


```python
total = 0
for sale in sales:
    total += sale
print(total)
```

## 3. Add a value


```python
sales.append(105)
print(sales)
```

## 4. Build a filtered list


```python
high_sales = []
for sale in sales:
    if sale >= 100:
        high_sales.append(sale)
print(high_sales)
```

## 5. AI Verify


```python
def remove_low_scores(scores, threshold):
    for score in scores:
        if score < threshold:
            scores.remove(score)
    return scores

scores = [2, 3, 8, 9, 7, 9]
print(remove_low_scores(scores, 5))
```

Correct cleaned list: `[8, 9, 7, 9]`.
