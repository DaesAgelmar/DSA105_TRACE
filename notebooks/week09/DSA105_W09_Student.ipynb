## 2. Branch and company totals


```python
company_total = 0

for branch in range(1, 3):
    branch_total = 0

    for day in range(1, 4):
        sale = float(input("Sale: "))
        branch_total += sale

    print("Branch total:", branch_total)
    company_total += branch_total

print("Company total:", company_total)
```

## 4. AI Verify


```python
company_total = 0

for branch in range(1, 3):
    branch_total = 0

    for day in range(1, 4):
        sale = float(input("Sale: "))
        branch_total += sale

    company_total = branch_total

print("Company total:", company_total)
```
