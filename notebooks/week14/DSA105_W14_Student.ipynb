## 3. AI Verify — two independent defects


```python
def average_adjusted_price(prices, threshold, discount_rate):
    adjusted = []

    for price in prices:
        if price >= threshold:
            adjusted.append(price - discount_rate)
        else:
            adjusted.append(price)

    total = 0
    for i in range(len(adjusted) - 1):
        total += adjusted[i]

    return total / len(adjusted)

prices = [80, 120, 200, 100]
print(average_adjusted_price(prices, 100, 0.10))
```

Intended rule:
- discount only values **strictly above** 100;
- discount is **10%**, not 0.10 currency units;
- include every adjusted value in the average.
