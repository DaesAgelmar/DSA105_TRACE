## 3. AI Verify


```python
def safe_divide(a, b):
    try:
        return a / b
    except:
        return 0

print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide(10, "two"))
```

Why is returning `0` for all failure types misleading? Repair by catching only the intended exception.
