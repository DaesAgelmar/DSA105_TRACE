### 2. Inspect raw lines


```python
with open("customers.txt", "r", encoding="utf-8") as file:
    raw_lines = file.readlines()

print(raw_lines)

for raw_line in raw_lines:
    print(repr(raw_line))
```

`repr()` makes hidden characters visible.


### 3. A reusable transformation


```python
def clean_lines(lines):
    """Return stripped, non-empty lines."""
    cleaned_lines = []

    for line in lines:
        cleaned_line = line.strip()

        if cleaned_line:
            cleaned_lines.append(cleaned_line)

    return cleaned_lines
```

The function knows nothing about files. That makes it easy to test with an ordinary list of strings.


```python
test_lines = ["Ada\n", "  Mert  \n", "\n"]
print(clean_lines(test_lines))
```

### 4. Parse numeric text


```python
def parse_amounts(lines):
    """Return floating-point amounts parsed from text lines."""
    amounts = []

    for line in lines:
        amounts.append(float(line))

    return amounts
```
