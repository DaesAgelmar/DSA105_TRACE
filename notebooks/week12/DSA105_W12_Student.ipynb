## 2. Read raw lines


```python
with open("customers.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

print(lines)
for line in lines:
    print(repr(line))
```

## 3. Clean


```python
names = []
for line in lines[1:]:
    names.append(line.strip())
print(names)
```

## 4. AI Verify


```python
def load_customer_names(filepath):
    with open(filepath, "r", encoding="utf-8") as file:
        lines = file.readlines()

    names = []
    for line in lines[1:]:
        names.append(line)

    return names

names = load_customer_names("customers.txt")
print(names)
print(repr(names[0]))
print(names[0] == "Ayşe")
```
