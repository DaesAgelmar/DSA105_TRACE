## 4. Multi-way decision

A customer receives a discount category based on purchase amount.


```python
purchase = 1750

if purchase >= 3000:
    print("Gold discount")
elif purchase >= 1500:
    print("Silver discount")
else:
    print("Standard pricing")
```
