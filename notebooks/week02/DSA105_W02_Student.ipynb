## 2. Algorithm step → Python statement

Some algorithm steps become visible Python statements very directly.

| Algorithmic intention | Python representation |
|---|---|
| Display a heading | `print("Three-Day Sales Report")` |
| Calculate a total | an arithmetic expression |
| Display a result | `print(...)` |
| Continue to next step | next line executes |

### Important
Python normally executes a simple program **from top to bottom**, one statement after another.

## 4. A first look at Python surface grammar

For today, notice only these visible elements:

- `print` asks Python to produce visible output;
- parentheses `(...)` contain what `print` should display;
- quotation marks `"..."` mark text;
- each statement must be written precisely enough for Python to parse it.
