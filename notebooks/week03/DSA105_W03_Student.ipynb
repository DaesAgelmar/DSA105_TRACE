## 3. Python data types

For this course, begin with four common types:

| Type | Typical meaning | Example |
|---|---|---|
| `int` | whole number | `1042` |
| `float` | decimal number | `425.50` |
| `str` | text | `"Data Analytics Handbook"` |
| `bool` | logical state | `True` |

Python can tell us the type of a value using `type()`.


### ACTIVITY 2 — Predict the types

Predict before running:

````python
print(type(product_name))
print(type(quantity))
print(type(unit_price))
print(type(is_member))
print(type(order_number))
````
